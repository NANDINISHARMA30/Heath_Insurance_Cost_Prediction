Data PreProcessing

Encoding the categorical values to num form

```python
df.replace({'sex':{'male':0,'female':1}},inplace=True)
df.replace({'smoker':{'no':0,'yes':1}},inplace=True)
df.replace({'region':{'southwest':0,'southeast':1,'northeast':2,'northwest':3}},inplace=True)
df
```

Splitting the features (x) and the target(y)

```python
x = df.drop(columns=['charges'], axis=1)  # Use 'columns' instead of 'column'
y = df['charges']  # Use bracket notation to select a column
```

```python
x
```

```python
y
```

Splitting the data into training and testing data

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
print(x.shape,x_train.shape,x_test.shape)
```

MODEL TRAINING

```python
lr=LinearRegression()
```

```python
lr.fit(x_train,y_train)
```

MODEL EVALUATION

```python
predicted_data=lr.predict(x_train)
```

```python
r2_train = metrics.r2_score(y_train, predicted_data)
print('R squared vale : ', r2_train)
```

```python
tested_data=lr.predict(x_test)
```

```python
r2_test = metrics.r2_score(y_test, tested_data)
print('R squared vale : ', r2_test)
```

PREDICTIVE SYSTEM

```python
input_data = (19,1,27.9,0,1,0)

# changing input_data to a numpy array
input_data_as_numpy_array = np.asarray(input_data)

# reshape the array
input_data_reshaped = input_data_as_numpy_array.reshape(1,-1)

prediction = lr.predict(input_data_reshaped)
print(prediction)

print('The insurance cost is USD ', prediction[0])
```
